## 1. Python vs C++: The Big Picture

Python is interpreted and dynamically typed: you write code and the interpreter runs it line by line, figuring out types as it goes. C++ is compiled and statically typed: the compiler checks your whole program and translates it into machine code before anything runs. The payoff is speed and control; the cost is that you must be more precise.

| Aspect | Python | C++ |
| :--- | :--- | :--- |
| **Execution** | Interpreted, line by line | Compiled to machine code first |
| **Types** | Dynamic: `x = 5` then `x = "hi"` is fine | Static: `int x = 5;` stays an int |
| **Blocks** | Indentation | Curly braces `{ }` |
| **Statements end with** | Newline | Semicolon `;` |
| **Entry point** | Top of the file runs | `int main()` runs |
| **Memory** | Managed automatically | You control it (more from Week 3) |
| **Speed** | Slower | Typically 10–100x faster |
| **Errors** | Mostly found when that line runs | Many found at compile time |

## 2. The Compilation Model

### 2.1 Your First Program

Here is a minimal C++ program. You can save this to `hello.cpp`.


````python
%%writefile hello.cpp
#include <iostream>

int main() {
    std::cout << "Hello, FundaClass!" << std::endl;
    return 0;
}
````

**Line by line:**
* `#include <iostream>` pastes the contents of the input/output library header into your file. Similar in spirit to `import` in Python.
* `int main()` is where execution begins. Every C++ program has exactly one `main`. It returns an `int` to the operating system.
* `std::cout` is the standard output stream; `<<` sends data into it. `std::` means "from the standard library namespace".
* `std::endl` ends the line and flushes the output. You can also write `"\n"`, which is faster.
* `return 0;` tells the OS the program succeeded. In `main` this line is optional, but it is good to be explicit at first.

> ℹ️ **Python vs C++:** `print("Hello")` becomes `std::cout << "Hello" << std::endl;`. Also, strings use double quotes only: `'a'` in C++ is a single character (`char`), not a string.


### 2.2 Compiling and Running

You cannot run a `.cpp` file directly. First compile it using a compiler such as `g++` (GCC) or `clang++`:

```bash
g++ -std=c++20 -Wall -Wextra hello.cpp -o hello
./hello          # Linux / macOS
hello.exe        # Windows
```

| Flag | Meaning |
| :--- | :--- |
| `-std=c++20` | Use the C++20 language standard |
| `-Wall -Wextra` | Turn on helpful warnings. Always use these. Many bugs are caught here. |
| `-o hello` | Name the output executable `hello` |
| `-g` | Include debug information (used with `gdb`) |
| `-O2` | Optimise the code for speed (use for final builds) |


### 2.3 What the Compiler Actually Does

Compiling is really four stages:

| # | Stage | What happens |
| :-: | :--- | :--- |
| **1** | **Preprocessing** | Handles lines starting with `#`. `#include` copies header text into your file; `#define` does text substitution. |
| **2** | **Compilation** | Checks syntax and types, and turns each `.cpp` file into assembly. Compile errors appear here. |
| **3** | **Assembly** | Turns assembly into an object file (`.o`) of machine code. |
| **4** | **Linking** | Joins your object files with library code into one executable. "Undefined reference" errors appear here. |

You can inspect each stage individually in your terminal:
```bash
g++ -E hello.cpp -o hello.i   # 1. Preprocess only
g++ -S hello.cpp              # 2. Compile to assembly (hello.s)
g++ -c hello.cpp              # 3. Compile to object file (hello.o)
g++ hello.o -o hello          # 4. Link into executable
```


### 2.4 Headers and Multiple Files

Larger programs are split into header files (`.h` or `.hpp`, containing declarations) and source files (`.cpp`, containing definitions). A header tells other files "this function exists"; the source file provides the actual body.


````python
%%writefile mathutils.h
#pragma once
int square(int x);          // declaration only
````

````python
%%writefile mathutils.cpp
#include "mathutils.h"
int square(int x) { return x * x; }   // definition
````

````python
%%writefile main.cpp
#include <iostream>
#include "mathutils.h"

int main() { 
    std::cout << square(7) << "\n"; 
}
````

Build multi-file programs with:
```bash
g++ main.cpp mathutils.cpp -o app
```
> **Note:** `<angle brackets>` are for standard/system headers, while `"quotes"` are for your own project headers.


### 2.5 Three Kinds of Errors

| Type | When | Example |
| :--- | :--- | :--- |
| **Compile-time** | Compiler refuses to build | Missing `;`, wrong type, unknown name |
| **Link-time** | Linker cannot find a function body | Declared `square` but never defined it |
| **Runtime** | Program builds but misbehaves | Division by zero, crash, wrong answer |

> 💡 **Tip:** Read compiler errors from the very first error reported. Later errors are often cascading side effects of the first. Fix one, recompile, and repeat.

## 3. Static Typing and Fundamental Types

In C++, every variable has a type fixed at compile time. You must declare the type, and it can never change.

```cpp
int age = 17;
double pi = 3.14159;
char grade = 'A';
bool passed = true;

age = "seventeen";   // ERROR: cannot assign a string to an int
```

> ℹ️ **Python vs C++:** In Python, `age = 17` then `age = "seventeen"` is legal because the *name* can point to anything. In C++, the *variable itself* is a typed box of fixed size.


### 3.1 Fundamental Types

| Type | Typical size | Holds | Example |
| :--- | :--- | :--- | :--- |
| `bool` | 1 byte | `true` or `false` | `bool ok = true;` |
| `char` | 1 byte | A single character | `char c = 'x';` |
| `int` | 4 bytes | Whole numbers, about ±2.1 billion | `int n = 42;` |
| `long long` | 8 bytes | Large whole numbers, about ±9.2 × 10¹⁸ | `long long big = 1e12;` |
| `unsigned int` | 4 bytes | Non-negative whole numbers, 0 to ~4.3 billion | `unsigned int u = 5;` |
| `float` | 4 bytes | Decimals, ~7 digits of precision | `float f = 1.5f;` |
| `double` | 8 bytes | Decimals, ~15 digits of precision | `double d = 1.5;` |

Sizes can vary by platform; the ones above are typical on modern 64-bit systems. Check with `sizeof`:


````python
%%writefile types_size.cpp
#include <iostream>

int main() {
    std::cout << "int:       " << sizeof(int) << " bytes\n";
    std::cout << "long long: " << sizeof(long long) << " bytes\n";
    std::cout << "double:    " << sizeof(double) << " bytes\n";
    std::cout << "char:      " << sizeof(char) << " byte\n";
}
````

**Output:**
```text
int:       4 bytes
long long: 8 bytes
double:    8 bytes
char:      1 byte
```

> ℹ️ **Python vs C++:** Python `int` can grow arbitrarily large without limit. C++ `int` is fixed at 32 bits, so it can overflow. Use `long long` when numbers can exceed ~2 billion (factorials, large sums, big products).


### 3.2 Integer Overflow


````python
%%writefile overflow.cpp
#include <iostream>

int main() {
    int a = 2147483647;            // largest signed 32-bit int
    long long b = 2147483647;
    std::cout << a + 1 << "\n";   // undefined behaviour for signed int
    std::cout << b + 1 << "\n";   // 2147483648, correct
}
````

> ⚠️ **Watch out:** Signed integer overflow is **undefined behaviour**: the compiler is technically allowed to do anything. On many machines you see a negative number due to wraparound, but you must never rely on this!


### 3.3 Integer Division and the Modulus Operator

When both operands are integers, `/` performs integer division and discards the fraction.

```cpp
std::cout << 7 / 2   << "\n";   // 3     (int / int)
std::cout << 7 / 2.0 << "\n";   // 3.5   (one operand is double)
std::cout << 7 % 2   << "\n";   // 1     (remainder)
std::cout << -7 / 2  << "\n";   // -3    (truncates toward zero)
std::cout << -7 % 2  << "\n";   // -1
```

> ℹ️ **Python vs C++:** In Python, `7 / 2` is `3.5` and `7 // 2` is `3`. Also `-7 // 2` is `-4` (floors) and `-7 % 2` is `1`. In C++, `/` on ints truncates toward zero, and `%` takes the sign of the left operand. There is no `//` operator and no `**` operator (use `std::pow` from `<cmath>`).


### 3.4 Initialisation Styles

```cpp
int a = 5;      // copy initialisation
int b(5);       // direct initialisation
int c{5};       // brace initialisation (preferred in modern C++)
int d{};        // zero-initialised: d is 0

int e = 3.9;    // allowed, silently becomes 3 (data loss!)
int f{3.9};     // ERROR: narrowing conversion not allowed
```

Brace initialisation `{}` refuses conversions that lose information, catching subtle mistakes early. Prefer it.

> ⚠️ **Watch out:** A local variable that you do not initialise holds garbage memory. `int x; std::cout << x;` prints an unpredictable value. Python would raise a `NameError`; C++ happily gives you junk. Always initialise!


### 3.5 `const`

```cpp
const double PI = 3.14159;
const int MAX_STUDENTS = 60;

PI = 3.14;      // ERROR: cannot modify a const
```

Use `const` for any value that should never change. It documents your intent and lets the compiler catch accidental modifications. `constexpr` is a stronger form where the value must be strictly known at compile time.


### 3.6 `auto`: Let the Compiler Deduce the Type

```cpp
auto n = 10;          // int
auto x = 3.14;        // double
auto big = 10LL;      // long long
auto ok = (n > 5);    // bool
auto c = 'z';         // char
```

`auto` is **not** dynamic typing. The type is decided once at compile time from the initial value, and then fixed forever. It is handy when type names become verbose.

> ⚠️ **Watch out:** `auto s = "hello";` makes `s` a `const char*`, not a `std::string`. When you want a string, write `std::string s = "hello";` (include `<string>`).


### 3.7 Type Conversion (Casting)

```cpp
int a = 7, b = 2;
double wrong = a / b;                       // 3   (integer division first!)
double right = static_cast<double>(a) / b;  // 3.5

char ch = 'A';
int code = static_cast<int>(ch);            // 65 (ASCII)
char next = ch + 1;                         // 'B'
```

Use `static_cast<Type>(value)` for explicit, safe, and visible conversions.


### 3.8 Strings (A First Look)


````python
%%writefile strings_demo.cpp
#include <iostream>
#include <string>

int main() {
    std::string first = "Fun";
    std::string second = "daClass";
    std::string full = first + second;     // concatenation with +

    std::cout << full << "\n";             // FundaClass
    std::cout << full.length() << "\n";    // 10
    std::cout << full[0] << "\n";          // F
    full[0] = 'f';                         // strings are mutable in C++
    std::cout << full << "\n";             // fundaClass
}
````

> ℹ️ **Python vs C++:** Python strings are immutable; C++ `std::string` is mutable. Use `.length()` or `.size()` instead of `len()`. Indexing with `[]` does not check bounds (use `.at(i)` for bounds-checked access).

## 4. Input and Output

### 4.1 Output with `cout`

```cpp
int age = 17;
double gpa = 8.75;
std::cout << "Age: " << age << ", CGPA: " << gpa << "\n";
```
**Output:**
```text
Age: 17, CGPA: 8.75
```
Chain values with `<<`. While C++ does not have Python's f-strings directly in standard stream syntax, chaining achieves the same result.


### 4.2 Input with `cin`


````python
%%writefile cin_demo.cpp
#include <iostream>

int main() {
    int a, b;
    std::cout << "Enter two numbers: ";
    std::cin >> a >> b;                 // reads whitespace-separated values
    std::cout << "Sum = " << a + b << "\n";
}
````

> ℹ️ **Python vs C++:** `cin >> a` reads a value directly into the type of `a`. You do not need to call `int(input())`. The static type of the variable decides how incoming text is parsed.


### 4.3 Reading a Whole Line with `getline`

`cin >>` stops at whitespace, so it cannot read a full name like `"Asha Verma"`. Use `std::getline`:


````python
%%writefile getline_demo.cpp
#include <iostream>
#include <string>

int main() {
    std::string name;
    int age;

    std::cout << "Enter age: ";
    std::cin >> age;

    std::cin.ignore();                  // discard leftover newline
    std::cout << "Enter full name: ";
    std::getline(std::cin, name);

    std::cout << name << " is " << age << " years old.\n";
}
````

> ⚠️ **Watch out:** After `cin >> age`, the Enter key's newline remains in the input buffer. If you call `getline` right after, it reads that empty leftover line immediately. Clear it with `std::cin.ignore()` first.


### 4.4 Formatting Decimals


````python
%%writefile format_demo.cpp
#include <iostream>
#include <iomanip>

int main() {
    double x = 3.14159265;
    std::cout << x << "\n";                                         // 3.14159
    std::cout << std::fixed << std::setprecision(2) << x << "\n";  // 3.14
    std::cout << std::setw(8) << x << "\n";                        //     3.14
}
````

`std::fixed` with `std::setprecision(n)` prints exactly $n$ digits after the decimal point. `std::setw(n)` sets the minimum field width. These require `#include <iomanip>`.


### 4.5 About `using namespace std;`

Many beginner tutorials write `using namespace std;` so they can type `cout` instead of `std::cout`. It is fine for tiny toy scripts, but in larger codebases it can cause naming collisions and ambiguities. In production and in this course, write `std::` explicitly so you know exactly where identifiers originate.

## 5. Control Flow

### 5.1 Operators

| Purpose | Python | C++ |
| :--- | :--- | :--- |
| **And / Or / Not** | `and`, `or`, `not` | `&&`, `\|\|`, `!` |
| **Equal / Not equal** | `==`, `!=` | `==`, `!=` |
| **Increment** | `x += 1` | `x += 1`, `x++`, `++x` |
| **Ternary** | `a if cond else b` | `cond ? a : b` |
| **Power** | `x ** 2` | `std::pow(x, 2)` or `x * x` |


### 5.2 `if` / `else if` / `else`


````python
%%writefile grade_checker.cpp
#include <iostream>

int main() {
    int marks;
    std::cout << "Enter marks: ";
    std::cin >> marks;

    if (marks >= 90) {
        std::cout << "Grade A\n";
    } else if (marks >= 75) {
        std::cout << "Grade B\n";
    } else if (marks >= 50) {
        std::cout << "Grade C\n";
    } else {
        std::cout << "Fail\n";
    }
}
````

> ℹ️ **Python vs C++:** The condition must go in parentheses `( )`, blocks require curly braces `{ }`, and it is `else if`, not `elif`. Indentation is for humans; the compiler ignores whitespace.
>
> ⚠️ **Watch out:** `if (x = 5)` assigns 5 to `x` and evaluates as true. You meant `if (x == 5)`. Turn on `-Wall` and the compiler will warn you!


### 5.3 `switch`

```cpp
int day = 3;
switch (day) {
    case 1:  std::cout << "Mon\n"; break;
    case 2:  std::cout << "Tue\n"; break;
    case 3:  std::cout << "Wed\n"; break;
    default: std::cout << "Other\n";
}
```
**Output:** `Wed`

`switch` works with integral and character types, not strings. Each case requires a `break;`, otherwise execution falls through into subsequent cases.


### 5.4 `for` Loops

```cpp
// Classic form: (init; condition; update)
for (int i = 0; i < 5; i++) {
    std::cout << i << " ";
}
// 0 1 2 3 4

// Counting down by 2
for (int i = 10; i > 0; i -= 2) {
    std::cout << i << " ";
}
// 10 8 6 4 2

// Range-based for (like Python's "for x in ...")
for (int x : {10, 20, 30}) {
    std::cout << x << " ";
}
// 10 20 30
```

> ℹ️ **Python vs C++:** `for i in range(5)` becomes `for (int i = 0; i < 5; i++)`. The upper bound and condition are explicit. The range-based `for (int x : collection)` is closest to Python's `for x in collection`.


### 5.5 `while` and `do-while`

```cpp
// while: check first
int n = 5, sum = 0;
while (n > 0) {
    sum += n;
    n--;
}
std::cout << sum << "\n";    // 15

// do-while: run at least once, then check
int choice;
do {
    std::cout << "Enter a number between 1 and 10: ";
    std::cin >> choice;
} while (choice < 1 || choice > 10);
```

`do-while` has no direct Python counterpart. It is ideal for menus and input validation because the loop body executes at least once before testing the condition.


### 5.6 `break` and `continue`

```cpp
for (int i = 1; i <= 10; i++) {
    if (i % 2 == 0) continue;   // skip even numbers
    if (i > 7) break;           // stop the loop entirely
    std::cout << i << " ";
}
// 1 3 5 7
```


### 5.7 Worked Example: FizzBuzz


````python
%%writefile fizzbuzz.cpp
#include <iostream>

int main() {
    for (int i = 1; i <= 15; i++) {
        if (i % 15 == 0)      std::cout << "FizzBuzz\n";
        else if (i % 3 == 0)  std::cout << "Fizz\n";
        else if (i % 5 == 0)  std::cout << "Buzz\n";
        else                  std::cout << i << "\n";
    }
}
````

### 5.8 Worked Example: Prime Checker


````python
%%writefile prime_checker.cpp
#include <iostream>

int main() {
    int n;
    std::cout << "Enter a number: ";
    std::cin >> n;

    bool isPrime = (n >= 2);
    for (int i = 2; i * i <= n; i++) {
        if (n % i == 0) {
            isPrime = false;
            break;
        }
    }
    std::cout << n << (isPrime ? " is prime\n" : " is not prime\n");
}
````

Notice `i * i <= n`: we only need to test divisors up to $\sqrt{n}$.


### 5.9 Scope

```cpp
int main() {
    int x = 10;
    {
        int y = 20;             // y exists only inside these braces
        std::cout << x + y;     // OK
    }
    // std::cout << y;          // ERROR: y is out of scope

    for (int i = 0; i < 3; i++) { }
    // std::cout << i;          // ERROR: i lived only inside the loop
}
```

> ℹ️ **Python vs C++:** In Python, a variable created inside a loop or `if` block remains accessible afterwards. In C++, a variable lives strictly inside the `{ }` block where it was declared.

## 6. Functions

### 6.1 Defining and Calling


````python
%%writefile functions_demo.cpp
#include <iostream>
#include <string>

int square(int x) {
    return x * x;
}

void greet(const std::string& name) {
    std::cout << "Hello, " << name << "!\n";
}

int main() {
    std::cout << square(6) << "\n";   // 36
    greet("Asha");
}
````

Syntax: `returnType name(paramType param, ...) { body }`.
* A function returning nothing uses `void`.
* Every parameter needs an explicit type, as does the return value.

> ℹ️ **Python vs C++:** There is no `def` and no `None`. A non-void function must return a value on every possible branch; failing to do so triggers undefined behaviour.


### 6.2 Declaration Before Use

C++ compilers parse top to bottom. A function must be declared before it can be called. You can either define it above `main`, or provide a prototype declaration above and define it below:


````python
%%writefile prototype_demo.cpp
#include <iostream>

int cube(int x);                 // prototype (declaration)

int main() {
    std::cout << cube(3) << "\n";   // 27
}

int cube(int x) {                // definition
    return x * x * x;
}
````

### 6.3 Default Arguments and Overloading


````python
%%writefile overloading_demo.cpp
#include <iostream>

// Default argument
double power(double base, int exp = 2) {
    double result = 1;
    for (int i = 0; i < exp; i++) result *= base;
    return result;
}

// Overloading: same name, different parameter types
int    maxOf(int a, int b)       { return a > b ? a : b; }
double maxOf(double a, double b) { return a > b ? a : b; }

int main() {
    std::cout << power(5) << "\n";       // 25   (exp defaults to 2)
    std::cout << power(2, 10) << "\n";   // 1024
    std::cout << maxOf(3, 8) << "\n";    // 8    (int version)
    std::cout << maxOf(2.5, 1.5) << "\n";// 2.5  (double version)
}
````

Overloading lets multiple functions share the same name; the compiler selects the correct overload based on argument types. Default parameters must appear at the end of the parameter list.


### 6.4 Pass by Value

By default, C++ passes arguments by value (copy). Modifications inside the function do not affect the caller's variable.


````python
%%writefile pass_by_val.cpp
#include <iostream>

void swapByValue(int a, int b) {
    int temp = a;
    a = b;
    b = temp;                    // only local copies swapped
}

int main() {
    int x = 1, y = 2;
    swapByValue(x, y);
    std::cout << x << " " << y << "\n"; // Output: 1 2
}
````

### 6.5 Pass by Reference

Add `&` after the type to create a reference parameter: an alias directly bound to the caller's actual variable.


````python
%%writefile pass_by_ref.cpp
#include <iostream>

void swapByRef(int& a, int& b) {
    int temp = a;
    a = b;
    b = temp;                    // swaps caller's actual variables
}

int main() {
    int x = 1, y = 2;
    swapByRef(x, y);
    std::cout << x << " " << y << "\n"; // Output: 2 1
}
````

### 6.6 Pass by Const Reference

Copying large objects (long strings, big vectors/containers) incurs overhead. When you only need read access, pass by `const &`: zero copying, with compile-time immutability guaranteed.


````python
%%writefile const_ref.cpp
#include <iostream>
#include <string>

int countVowels(const std::string& s) {     // no copy, read-only
    int count = 0;
    for (char c : s) {
        if (c=='a' || c=='e' || c=='i' || c=='o' || c=='u') count++;
    }
    return count;
}

int main() {
    std::cout << countVowels("FundaClass Academy") << "\n";
}
````

| Style | Syntax | When to Use |
| :--- | :--- | :--- |
| **By value** | `int x` | Small primitive types (`int`, `double`, `char`, `bool`) or when an independent copy is required |
| **By reference** | `int& x` | The function must mutate the caller's variable in place |
| **By const reference** | `const std::string& s` | Large objects where you only need read access (strings, vectors, custom structs) |

> ℹ️ **Python vs C++:** In Python, every argument is passed by object reference, but reassigning a variable name inside a function does not mutate caller primitives. In C++, you explicitly choose between passing a copy or a direct reference.


### 6.7 Returning Results: One Value or Multiple

A function returns a single value. To return multiple results without custom structs, use reference parameters as *output parameters*:


````python
%%writefile output_params.cpp
#include <iostream>

void divide(int a, int b, int& quotient, int& remainder) {
    quotient  = a / b;
    remainder = a % b;
}

int main() {
    int q, r;
    divide(17, 5, q, r);
    std::cout << "q=" << q << ", r=" << r << "\n"; // q=3, r=2
}
````

*(Note: In later weeks you will also learn `std::pair`, `std::tuple`, and structured bindings).*


### 6.8 Recursion


````python
%%writefile recursion_demo.cpp
#include <iostream>

long long factorial(int n) {
    if (n <= 1) return 1;            // base case
    return n * factorial(n - 1);     // recursive case
}

int main() {
    std::cout << factorial(5)  << "\n";    // 120
    std::cout << factorial(20) << "\n";    // 2432902008176640000
}
````

The return type is `long long` because `20!` easily exceeds a 32-bit `int`. Even `long long` overflows at $21!$, illustrating the fixed bounds of C++ numeric types.


### 6.9 Worked Example: Menu-Driven Calculator


````python
%%writefile calculator.cpp
#include <iostream>

double add(double a, double b) { return a + b; }
double sub(double a, double b) { return a - b; }
double mul(double a, double b) { return a * b; }

bool divide(double a, double b, double& result) {
    if (b == 0) return false;        // signal failure
    result = a / b;
    return true;
}

int main() {
    double x, y;
    char op;
    std::cout << "Enter: number operator number (e.g. 8 / 2): ";
    std::cin >> x >> op >> y;

    switch (op) {
        case '+': std::cout << add(x, y) << "\n"; break;
        case '-': std::cout << sub(x, y) << "\n"; break;
        case '*': std::cout << mul(x, y) << "\n"; break;
        case '/': {
            double r;
            if (divide(x, y, r)) std::cout << r << "\n";
            else std::cout << "Error: division by zero\n";
            break;
        }
        default:  std::cout << "Unknown operator\n";
    }
}
````

## 7. Common Mistakes and How to Fix Them

| Mistake | Symptom | Fix |
| :--- | :--- | :--- |
| **Missing `;`** | Compile error, often reported on the *next* line | Add the semicolon |
| **Forgot `#include`** | `"cout was not declared"` | Include `<iostream>`, `<string>`, etc. |
| **Forgot `std::`** | `"cout was not declared"` | Write `std::cout` |
| **`=` instead of `==`** | Condition always evaluates to true or false | Use `==`; compile with `-Wall` |
| **Uninitialised variable** | Random garbage values | Always provide an initial value (`int x{0};`) |
| **Integer division** | `7 / 2` gives `3` | Make one operand float/double: `7.0 / 2` or `static_cast<double>(a) / b` |
| **Overflow in `int`** | Negative or wrapped large numbers | Use `long long` |
| **Missing `break` in `switch`** | Several cases run consecutively | Add `break;` |
| **Missing `return`** | Garbage return result | Return explicitly on every branch; check compiler warnings |
| **Off-by-one in loops** | Loop runs one iteration too many or few | Check `<` vs `<=` and starting bound index |

### **Key Takeaways**
1. **Compilation pipeline:** Source → Preprocess → Compile → Assemble → Link → Executable.
2. **Fixed typing:** Types are determined at compile time. Choose carefully, and guard against integer overflow.
3. **Safety first:** Always initialise variables, and compile with `-Wall -Wextra`.
4. **Division nuance:** Integer division truncates towards zero; cast to `double` when fractions are required.
5. **Block scoping:** Variables live strictly inside the `{ }` block in which they are declared.
6. **Parameter passing:** Arguments are copied by default. Use `&` to modify caller values and `const &` to pass large structures efficiently without copying.
